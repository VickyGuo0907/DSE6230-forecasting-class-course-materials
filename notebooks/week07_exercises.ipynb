```python
import pandas as pd
import matplotlib.pyplot as plt
from neuralforecast import NeuralForecast
from neuralforecast.models import NBEATS, NHITS
from neuralforecast.utils import AirPassengersDF
```

### Exercise 1: Training N-BEATS

We will use the AirPassengersDF but treat it as a "panel" (Nixtla format requires `unique_id`, `ds`, `y`).

```python
Y_df = AirPassengersDF
Y_df.head()
```

```python
horizon = 12
models = [
    NBEATS(input_size=2 * horizon, h=horizon, max_steps=100),
    NHITS(input_size=2 * horizon, h=horizon, max_steps=100)
]

nf = NeuralForecast(models=models, freq='M')
nf.fit(df=Y_df)
forecasts = nf.predict()
forecasts.head()
```

### Exercise 2: Visualizing Results

Plot the N-BEATS and NHITS forecasts against the training history.

```python
pd.concat([Y_df, forecasts]).set_index('ds').groupby('unique_id').plot()
plt.show()
```
